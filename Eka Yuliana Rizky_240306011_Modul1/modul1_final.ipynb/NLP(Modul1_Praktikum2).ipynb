```python
## B. Import Library
import pandas as pd
import nltk
from pathlib import Path
from nltk.tokenize import sent_tokenize, word_tokenize

nltk.download("punkt")
nltk.download("punkt_tab")

print("Library siap digunakan.")
```

```python
## C. Load Dataset
candidate_paths = [
    Path("../dataset/dataset.csv"),
    Path("dataset.csv"),
    Path("../dataset.csv"),
    Path("/mnt/data/dataset.csv")
]

dataset_path = next((p for p in candidate_paths if p.exists()), None)
if dataset_path is None:
    raise FileNotFoundError(
        "Dataset tidak ditemukan. Letakkan dataset.csv di folder dataset "
        "atau sesuaikan candidate_paths."
    )

data = pd.read_csv(dataset_path)
text_column = "comment"

print("Lokasi dataset:", dataset_path)
print("Ukuran dataset:", data.shape)
print("Kolom:", data.columns.tolist())
display(data.head())
```

```python
## D. Pemeriksaan dan Persiapan Data
print("Jumlah baris awal:", len(data))
print("Missing value pada kolom teks:", data[text_column].isna().sum())
print("Duplikat teks sebelum cleaning:", data[text_column].dropna().duplicated().sum())

data_teks = data.dropna(subset=[text_column]).copy()
data_teks[text_column] = data_teks[text_column].astype(str)
data_teks = data_teks.drop_duplicates(subset=[text_column]).copy()
data_teks = data_teks.reset_index(drop=True)

print("Jumlah komentar setelah menghapus nilai kosong dan duplikat:", len(data_teks))
display(data_teks[[text_column]].head())
```

```python
## E. Sentence Segmentation
text = (
    "Saya belajar NLP. "
    "NLP sangat menarik. "
    "Saya ingin menjadi NLP Engineer."
)

sentences = sent_tokenize(text)
for i, sentence in enumerate(sentences, start=1):
    print(f"Kalimat {i}: {sentence}")
print("Jumlah kalimat:", len(sentences))
```

```python
## F. Word Tokenization
text = "Saya belajar Natural Language Processing."
tokens = word_tokenize(text)

print("Teks asli:", text)
print("Hasil tokenisasi:", tokens)
print("Jumlah token:", len(tokens))
```

```python
## G. Character Tokenization
text = "NLP"
characters = list(text)

print("Teks asli:", text)
print("Hasil character tokenization:", characters)
print("Jumlah karakter:", len(characters))
```

```python
## H. Tokenisasi Dataset
data_teks["tokens"] = data_teks[text_column].apply(word_tokenize)
data_teks["jumlah_token"] = data_teks["tokens"].apply(len)

display(data_teks[[text_column, "tokens", "jumlah_token"]].head(10))
```

```python
## I. Statistik Jumlah Token
print("Statistik jumlah token per komentar:")
display(data_teks["jumlah_token"].describe())

print("Jumlah dokumen setelah cleaning:", len(data_teks))
print("Total token:", data_teks["jumlah_token"].sum())
print("Rata-rata token per dokumen:", round(data_teks["jumlah_token"].mean(), 2))
```

```python
## J. Perbandingan Jumlah Kata dan Jumlah Token
data_teks["jumlah_kata"] = data_teks[text_column].apply(lambda x: len(x.split()))

display(data_teks[[text_column, "jumlah_kata", "jumlah_token"]].head(10))
display(data_teks[["jumlah_kata", "jumlah_token"]].describe())
```

```python
## K. Eksperimen Tokenisasi
texts = [
    "Saya suka produk ini.",
    "Bagus banget!!!",
    "Tidak bagus...",
    "Harga Rp50.000.",
    "email@example.com",
    "https://example.com",
    "yg bagus bgt sih",
    "aku udah beli nih kak"
]

for sample in texts:
    print("TEXT :", sample)
    print("TOKEN:", word_tokenize(sample))
    print("JUMLAH TOKEN:", len(word_tokenize(sample)))
    print("-" * 50)
```

```python
## L. Perbandingan Tokenizer: NLTK dan split()
comparison_texts = [
    "Saya belajar NLP.",
    "Bagus banget!!!",
    "Harga Rp50.000",
    "yg bagus bgt sih",
    "https://example.com"
]

comparison_rows = []
for sample in comparison_texts:
    nltk_tokens = word_tokenize(sample)
    split_tokens = sample.split()
    comparison_rows.append({
        "Teks": sample,
        "NLTK word_tokenize": nltk_tokens,
        "Jumlah token NLTK": len(nltk_tokens),
        "split()": split_tokens,
        "Jumlah hasil split": len(split_tokens)
    })

comparison_df = pd.DataFrame(comparison_rows)
display(comparison_df)
```

```python
## M. Tokenisasi Karakter pada Dataset
if not data_teks.empty:
    sample_text = data_teks[text_column].iloc[0]
    characters = list(sample_text)
    print("Contoh komentar:", sample_text)
    print("Karakter:", characters)
    print("Jumlah karakter:", len(characters))
else:
    print("Tidak ada komentar yang dapat diproses.")
```
